The results indicate that the baseline model is able to detect objects in UAV imagery, but its overall detection performance remains limited, particularly for small, crowded, and less-represented object categories.

## 11.3 Training Behaviour

The training and validation loss curves show an overall decreasing trend throughout the 20 training epochs.

The training box loss, classification loss, and DFL loss progressively decreased during training. The validation losses also generally decreased, although some fluctuations were observed during training.

The continued reduction in training losses indicates that the model was learning useful detection features from the dataset. However, the relatively low final mAP values indicate that lower training loss did not translate into highly accurate object localization and classification across all object categories.

## 11.4 Detection Performance Analysis

The detection-performance curves show that precision generally improved as the confidence threshold increased, while recall decreased as increasingly strict confidence thresholds were applied.

The F1-confidence curve indicates an overall maximum F1 score of approximately 0.25 at a confidence threshold of approximately 0.128.

This indicates that a relatively low confidence threshold provides a better balance between precision and recall for this baseline model. However, visual inspection of predictions at confidence = 0.13 showed a substantial increase in weak and duplicate detections in crowded scenes.

Therefore, the confidence threshold must be selected based not only on the F1 curve but also on the practical quality of the resulting detections.

## 11.5 Confusion Matrix Analysis

The normalized confusion matrix indicates substantial variation in detection performance between object classes.

The Car class shows comparatively stronger detection behaviour than several smaller or less-represented classes. In contrast, classes such as Pedestrian, People, Bicycle, Tricycle, Awning Tricycle, Bus, and Motor show considerable confusion with other classes and/or background.

A large proportion of objects are also associated with the background category, indicating missed detections.

This behaviour is consistent with the challenges of UAV-based object detection, where objects can occupy only a small number of pixels and may be partially occluded or densely packed.

## 11.6 Class-Level Challenges

The principal challenges identified from the evaluation are:

1. **Small objects** — many objects occupy relatively small regions of the UAV image, making localization difficult.
2. **Class imbalance** — the dataset contains substantially more examples of some classes than others.
3. **Crowded scenes** — densely populated traffic and pedestrian scenes result in overlapping objects and duplicate detections.
4. **Object occlusion** — partially visible objects are difficult to localize and classify accurately.
5. **Low-confidence detections** — weaker object classes frequently produce predictions with low confidence scores.

These factors contribute directly to reduced recall, precision, and mAP.

## 11.7 Unseen-Image Evaluation

The trained model was additionally evaluated on five images that were not used during the training or validation process.

The model successfully produced detections across different UAV viewing conditions, including:

- aerial views of parked vehicles,
- road traffic scenes,
- dense pedestrian environments,
- residential/urban areas,
- high-rise urban environments.

The unseen-image results demonstrate that the model has learned generalizable visual features rather than simply reproducing the training examples.

However, the predictions also reveal several practical limitations, including duplicate detections, low-confidence predictions, and difficulty with small or heavily crowded objects.

## 11.8 Confidence Threshold Experiment

An additional inference experiment was conducted using a confidence threshold of 0.13.

Although the lower threshold allowed the model to produce more detections, visual inspection showed an increase in weak and overlapping predictions, particularly in crowded pedestrian and traffic scenes.

Therefore, reducing the confidence threshold does not automatically improve practical detection quality.

The experiment demonstrates the trade-off between detection sensitivity and prediction reliability.

## 11.9 Overall Assessment

The YOLO11n baseline provides a functional starting point for UAV-based object detection. The model successfully detects several object categories in unseen aerial imagery, with comparatively stronger performance for larger and more frequently represented objects.

Nevertheless, the relatively low mAP@50 of 0.1903 and mAP@50-95 of 0.1031 demonstrate that substantial improvement is required before the model can be considered suitable for high-reliability deployment.

The evaluation provides a clear baseline against which future improvements can be measured.

## 11.10 Limitations

The major limitations identified in the present experiment are:

- Limited training duration of 20 epochs.
- Small YOLO11n model capacity.
- Strong class imbalance.
- Large number of small objects.
- Dense and overlapping objects in UAV imagery.
- Difficulty detecting minority classes.
- Limited recall at the selected operating point.
- Duplicate detections in crowded scenes.
- Only five unseen images were used for qualitative testing.

These limitations should be considered when interpreting the reported performance.

## 11.11 Future Improvements

Future work could investigate:

- Larger YOLO architectures such as YOLO11s or YOLO11m.
- Higher image resolution.
- Improved class balancing.
- Targeted augmentation for small objects.
- Mosaic and multi-scale training strategies.
- Improved annotation quality.
- More training epochs with appropriate early stopping.
- Class-specific confidence thresholds.
- Improved NMS/IoU settings.
- Additional unseen test images.
- Comparison against other object-detection architectures.

7. **Minority classes**
